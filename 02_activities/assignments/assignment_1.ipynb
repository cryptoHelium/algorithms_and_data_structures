## Problem Three: Move All Zeros to End
**Description**  
Given a list of integers, move all zeros to the end while maintaining the relative order of the non-zero elements. 

**Examples**
* Example 1:
    ```python
    Input: nums = [0, 1, 0, 3, 12]
    Output: [1, 3, 12, 0, 0]
    ```
* Example 2:
    ```python
    Input: nums = [4, 0, 5, 0, 0, 6]
    Output: [4, 5, 6, 0, 0, 0]
    ```

**Problem 3 Starter Code**

````python
from typing import List

def move_zeros_to_end(nums: List[int]) -> List[int]:

    #Sort the list using in-house sort method for lists

    nums.sort()

    #Determine the length of the list

    length = len(nums)

    #now we iterate and find the zeros. use while loop
    #We need to track the items checked and we stop when the original list finishes

    items_checked = 0
    i = 0  #index

    while items_checked < length:
        # Check if the items equals 0
        if nums[i] == 0:
            # temporarily remove it from the list
            nums.pop(i) #pop the 0
            nums.append(0) #Add it to the end of the list
            # We dont need to increase the index here because the next items shifted left into index i when we popped
        else:
            i += 1 #in this case move to the next item if it wasn't 0.

        items_checked += 1 #Keep track of the progress through the original list

    return nums
````

````python
#Test code above
nums2 = [0, 1, 0, 3, 12]

print(move_zeros_to_end(nums2))

nums3 = [4, 0, 5, 0, 0, 6]

print (move_zeros_to_end(nums3))
````
