## Problem Three: Missing Number in Range
**Description**  
You are given a list containing `n` non-negative integers in the range `[1, n]`. Return a list of numbers that are missing from the range `[1, n]` in the given list. If there is no missing number, return -1. 

**Note**: integers in the list may not be unique.

**Examples**

* Example 1:

    Input: `lst = [1, 1, 3]`

    Output: [2]

* Example 2:

    Input: `lst = [5,3,2,4,1]`

    Output: -1

* Example 3:

    Input: `lst = [6, 8, 2, 3, 5, 7, 1, 1, 2, 10]`

    Output: [4, 9]

**Problem 3 Starter Code**

```python
from typing import List
def missing_num(nums: List) -> int:
  # Code your solution here #
  # TODO

  n = len(nums)

  #convert to set data structure. searching the set is faster. It is O(1). Doing this
  #allows my function to be O(N) The reason is I am doing O(1) work N times. If I used a list
  #to search I would maybe need nested for loops and it would then be O(n^2) I am avoiding this
  #by converting to a set

  nums_to_set = set(nums)

  #create list of missing numbers

  missing_numbers = []

  #iterate through set and check if all numbers in the range are in the set. 
  #this is faster than searching the list
  #if the number is missing, add it to the missing numbers list (missing_numbies)
  # if nothing is missing, return -1

  for i in range (1, n + 1):
    if i not in nums_to_set:
      missing_numbers.append(i)

  if not missing_numbers:
      return -1

  return missing_numbers

#Testing 
lst = [1,1,3]

#should print 2
print (missing_num(lst))

lst = [5,3,2,4,1]

#should pring -1
print (missing_num(lst))

lst = [6, 8, 2, 3, 5, 7, 1, 1, 2, 10]

#should pring 4,9
print (missing_num(lst))
```


4. Copy the solution your partner wrote. 


```python
# Your answer here

#Their code is below from github

def is_valid_brackets(s: str) -> bool:
    stack = []

    for bracket in s:
        if bracket in "([{":
            stack.append(bracket)
        else:
            if len(stack) == 0:
                return False

            last = stack.pop()

            if bracket == ")" and last != "(":
                return False
            if bracket == "]" and last != "[":
                return False
            if bracket == "}" and last != "{":
                return False

    return len(stack) == 0

#Testing the code:

print(is_valid_brackets("([]{})"))  # True, they proper nest and match
print(is_valid_brackets("([)]"))    # False, but here opening braket is [, but closing does not match.....so wrong
print(is_valid_brackets("()[]{}"))  # True, all matched
print(is_valid_brackets("[{]}"))    # False, again as 2nd option, opening bracket is {, but closing does not match, so wrong..
```
